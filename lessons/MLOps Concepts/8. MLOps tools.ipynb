Let's dive into some possible tools we can use per component.

## Feature store
Such **Feast, Hopsworks**
- Both open-source
- **Feast**: self-managed feature store, meaning we have to manage it ourselves, which requires more work but also provides more flexibility compared to other feature stores.
- **Hopsworks**: part of larger platform. It is, therefore, more likely to be used if the rest of the Hopsworks tools are already in use.

## Experiment tracking
We can use **MLFlow, ClearML,** and **Weights and Biases**
- **MLFlow and ClearML**: full machine learning lifecycle tools, including experiment tracking.
- **MLFlow**: is specialized on machine learning development.
- **ClearML**: also provides tools to deploy models.
- **Weights and Biases**: tracking and visualizing experiments results

## Containerization
For containerization, **Docker** is the most popular tool to containerize an application.

- **Docker**: containerizing applications
- **Kubernetes**: running containerized application, enabling automatic deployment and scalability.
- **Cloud providers**: provides Kubernetes-like services

## CI/CD pipeline
Tools such as **Jenkins** and **GitLab**
- **Jenkins**: open-source continuous integration tool
- **GitLab**: code sharing and version control through repositories

Both of these tools allow developers to work on code together using a repository. For each project, there is often a separate repository, which we can see as a directory that contains all code for the project.

## Monitoring
There is a broad range of tools for monitoring machine learning project.
- **Fiddler**: machine learning model monitoring, focuses on the model performance, for instance, how well our model predictions are doing.
- **Great expectations**: data monitoring, focuses on data monitoring, for instance, how much data is missing in a certain column.

Both **Fiddler** and **Great expectations** provide statistical monitoring tools.
